# Article Synthesis

The article *Data-Driven Minimum-Energy Controls for Linear Systems*, proposed by Baggio, Katewa, and Pasqualetti, addresses one of the most fundamental problems in modern control theory: the computation of minimum-energy control inputs for linear dynamical systems. More specifically, the work investigates how to drive a linear discrete-time system toward a desired final state while minimizing the control energy, but without requiring explicit knowledge of the system matrices.

This shift in perspective represents a major conceptual transition from traditional model-based control approaches toward modern data-driven methodologies. In classical control theory, the design of optimal controllers typically assumes that the system dynamics are fully known through the matrices:

$$
A
\quad \text{and} \quad
B,
$$

which define the linear system:

$$
x(t+1)=Ax(t)+Bu(t),
$$

where:

- $(x(t) \in \mathbb{R}^n)$ is the state vector,
- $(u(t) \in \mathbb{R}^m)$ is the control input vector,
- $(A)$ describes the internal dynamics of the system,
- $(B)$ describes how the control inputs affect the system dynamics.

Within this classical framework, the minimum-energy control problem consists of determining a sequence of control inputs capable of transferring the system from an initial state \(x_0\) to a desired final state \(x_f\) while minimizing the quadratic energy cost:

$$
\sum_{t=0}^{T-1} ||u(t)||_2^2.
$$

This problem plays a central role in several engineering and scientific domains, including:

- robotics,
- aerospace systems,
- biological networks,
- distributed systems,
- power systems,
- network control,
- and optimal resource allocation problems.

Traditionally, solving this problem requires explicit knowledge of the system dynamics. The optimal solution is typically computed using tools such as:

- controllability matrices,
- controllability Gramians,
- Moore-Penrose pseudoinverses,
- orthogonal projections,
- and spectral matrix decompositions.

However, the article starts from a crucial practical observation: in many real-world scenarios, accurate mathematical models of the system are either unavailable or extremely difficult to identify.

In complex dynamical systems, identifying the matrices $A$ and $B$ may be:

- computationally expensive,
- numerically unstable,
- highly sensitive to noise,
- or simply infeasible.

This issue becomes particularly severe in large-scale systems such as:

- power grids,
- biological systems,
- social networks,
- interconnected cyber-physical systems,
- and large distributed infrastructures.

In these scenarios, the effort required to identify an accurate model may become as difficult as the original control problem itself.

Motivated by this limitation, the article proposes a fundamentally different perspective: instead of identifying the model first and solving the control problem afterward, the authors investigate whether it is possible to compute minimum-energy controls directly from experimental data.

This idea constitutes the central contribution of the paper.

Rather than reconstructing the matrices $A$ and $B$, the proposed approach uses experimental trajectories obtained from the system itself. Each experiment consists of:

- applying an input sequence to the system,
- observing the resulting final state,
- and storing this information in structured data matrices.

Using these experimental datasets, the authors show that the optimal control input can be reconstructed directly as a linear combination of previously applied experimental inputs.

This reformulation transforms the problem into a deeply geometric and algebraic one. The solution no longer relies on explicit dynamical identification, but instead on:

- vector spaces,
- linear independence,
- rank conditions,
- kernels,
- images,
- orthogonal projections,
- pseudoinverses,
- and geometric properties of experimental subspaces.

One of the insights of the article is that the experimental inputs span a subspace inside the control signal space:

$$
\mathbb{R}^{mT},
$$

where:

- $(m)$ is the number of control inputs,
- $(T)$ is the control horizon.

The quality of the reconstructed controller depends directly on how well the experimental inputs geometrically cover this space.

Within this framework, the paper establishes several fundamental results regarding the minimum number of experiments required for exact reconstruction.

In particular, the authors show that:

$$
N \ge n
$$

is necessary to reach arbitrary final states, while:

$$
N = mT
$$

is sufficient to reconstruct the exact minimum-energy control input, provided that the experimental inputs are linearly independent.

This distinction reveals one of the deepest geometric insights of the article: reaching an arbitrary state is not equivalent to reconstructing the optimal control law associated with that state.

The first problem concerns the coverage of the state space:

$$
\mathbb{R}^n,
$$

whereas the second concerns the coverage of the entire control signal space:

$$
\mathbb{R}^{mT}.
$$

Another particularly important contribution of the paper is its geometric interpretation of minimum-energy control.

The authors show that the optimal control corresponds to the component orthogonal to the kernel of the controllability operator. Geometrically, any component of the control signal belonging to the kernel produces no effect on the final state while still consuming energy.

Consequently, the minimum-energy solution can be interpreted as the orthogonal projection of the control input onto the effective controllable subspace.

This interpretation establishes a connection between:

- optimal control,
- orthogonal projections,
- pseudoinverse theory,
- and geometric properties of linear operators.

Beyond its theoretical contributions, the article also investigates important numerical aspects of the problem. The authors analyze:

- numerical conditioning,
- pseudoinverse stability,
- and sensitivity to experimental richness.

The numerical experiments demonstrate that different mathematical formulations of the same control problem can exhibit significantly different numerical behaviors, especially in high-dimensional systems.

In this sense, the paper also connects data-driven control with:

- low-rank structures,
- conditioning theory,
- linear algebra,
- optimization,
- system identification,
- data-driven learning,
- and optimal control theory.

Its main contribution is demonstrating that essential dynamical and control properties can be inferred directly from experimental data without explicit model reconstruction.

Ultimately, the article provides a rigorous and elegant bridge between classical model-based control and modern data-driven methodologies, showing that minimum-energy control can be understood fundamentally as a geometric problem involving experimental subspaces, projections, and low-dimensional structures in high-dimensional control spaces.

# Linear Algebra Tools Used in the Article

The article *Data-Driven Minimum-Energy Controls for Linear Systems* is built upon concepts from linear algebra, geometric control theory, and numerical matrix analysis. The proposed methodology reformulates the minimum-energy control problem almost entirely in terms of vector spaces, linear operators, projections, rank conditions, and matrix factorizations.

Rather than relying on explicit identification of the system matrices, the article interprets the control problem geometrically through experimental subspaces generated from data. Consequently, linear algebra becomes the central mathematical language of the entire framework.

The main linear algebra tools used throughout the article are described below.

---

## 1. Vector Spaces and Subspaces

The entire formulation is based on the interpretation of:

- the state space,
- the control input space,
- and the experimental data space

as vector spaces.

The system state evolves in:

$$
\mathbb{R}^n,
$$

while the control input sequence belongs to:

$$
\mathbb{R}^{mT},
$$

where:

- \(n\) is the dimension of the state space,
- \(m\) is the number of control inputs,
- \(T\) is the control horizon.

The article investigates how experimental trajectories span subspaces inside these spaces and how these subspaces can be used to reconstruct optimal control signals.

Subspace geometry is therefore one of the foundational concepts of the paper.

---

## 2. Linear Dynamical Systems

The system dynamics are represented using the discrete-time linear state-space model:

$$
x(t+1)=Ax(t)+Bu(t),
$$

where:

$$
A\in\mathbb{R}^{n\times n},
\qquad
B\in\mathbb{R}^{n\times m}.
$$

This representation itself is fundamentally a linear algebraic formulation, where:

- \(A\) acts as a linear operator describing the autonomous dynamics;
- \(B\) maps control inputs into the state space.

The evolution of the system over time is described through repeated matrix multiplications and compositions of linear operators.

---

## 3. Controllability Matrix

One of the central tools used in the article is the controllability matrix:

$$
C_T=
\begin{bmatrix}
B & AB & A^2B & \cdots & A^{T-1}B
\end{bmatrix}.
$$

The controllability matrix describes how control inputs propagate through the system dynamics over the control horizon.

Geometrically, its columns span the reachable subspace of the system.

The article relies heavily on the rank and geometric properties of this matrix to characterize:

- reachability,
- controllability,
- and minimum-energy control solutions.

---

## 4. Rank and Linear Independence

Rank conditions play a fundamental role throughout the paper.

The article repeatedly analyzes whether collections of experimental signals are sufficient to span the relevant spaces.

In particular, conditions such as:

$$
\mathrm{rank}(C_T)=n
$$

are used to determine whether the system is controllable.

Similarly, the experimental input matrix:

$$
U=
\begin{bmatrix}
u_1 & \cdots & u_N
\end{bmatrix}
$$

must possess full row rank to guarantee that the experimental inputs span the entire control signal space.

Linear independence is therefore directly connected to:

- controllability,
- experimental richness,
- and exact reconstruction of optimal controls.

---

## 5. Kernel and Image of Linear Operators

The article makes extensive use of the concepts of:

- kernel (null space),
- image (range),
- and orthogonality.

The kernel of the controllability operator:

$$
\mathrm{Ker}(C_T)
$$

plays a particularly important role.

Geometrically, this kernel represents the set of control signals that produce no effect on the final state of the system.

The paper demonstrates that minimum-energy controls correspond to control vectors orthogonal to this kernel.

This interpretation establishes a direct geometric connection between:

- control optimality,
- orthogonal decomposition,
- and projection theory.

---

## 6. Moore-Penrose Pseudoinverse

The Moore-Penrose pseudoinverse is one of the most important algebraic tools used in the article.

It is denoted by:

$$
A^\dagger.
$$

The pseudoinverse is used to compute minimum-norm solutions of underdetermined linear systems.

In the context of the paper, it appears in expressions such as:

$$
u^*=C_T^\dagger x_f,
$$

and in the data-driven formulation:

$$
\hat u=UX^\dagger x_f.
$$

Geometrically, the pseudoinverse computes the orthogonal projection of a vector onto the image of a linear operator while simultaneously minimizing the Euclidean norm of the solution.

This property is fundamental for obtaining minimum-energy controls.

---

## 7. Orthogonal Projections

Orthogonal projection operators are central to the geometric interpretation developed in the article.

The paper explicitly constructs projections of the form:

$$
I-UK(UK)^\dagger,
$$

which remove components belonging to the kernel of the controllability operator.

Geometrically, these projections eliminate energetically useless control directions.

This interpretation allows the minimum-energy solution to be viewed as the orthogonal component of the control vector relative to the uncontrollable subspace.

---

## 8. Experimental Data Matrices

The article reformulates the control problem entirely in terms of experimental data matrices:

$$
U=
\begin{bmatrix}
u_1 & \cdots & u_N
\end{bmatrix},
$$

and:

$$
X=
\begin{bmatrix}
x_1 & \cdots & x_N
\end{bmatrix}.
$$

These matrices encode experimental trajectories and become the primary objects manipulated throughout the proposed data-driven framework.

The relationship:

$$
X=C_TU
$$

is one of the key equations of the paper and establishes a direct linear algebraic link between:

- experimental inputs,
- controllability,
- and reachable states.

## 10. Controllability Gramian

The controllability Gramian:

$$
W_T=C_TC_T^T
$$

is another important matrix-theoretic tool used in the article.

It appears in the classical model-based solution for minimum-energy control.

The Gramian encodes how easily different directions in the state space can be reached through control inputs.

Geometrically, it defines the energy metric associated with the reachable subspace.

The article compares the classical Gramian-based formulation with the proposed data-driven approaches.

---

## 11. Low-Rank Geometry and Experimental Subspaces

A major conceptual contribution of the article is the interpretation of experimental trajectories as generators of low-dimensional subspaces inside the control signal space.

The paper essentially reformulates control synthesis as a subspace reconstruction problem.

This perspective connects the work to broader areas such as:

- subspace identification,
- low-rank approximation,
- and data-driven dynamical systems.


## Overall Interpretation

The article demonstrates that minimum-energy control can be reformulated almost entirely as a geometric linear algebra problem.

Rather than explicitly identifying the system matrices, the proposed framework reconstructs optimal controls through:

- experimental subspaces,
- orthogonal projections,
- pseudoinverses,
- and rank conditions.

Consequently, the paper establishes a deep connection between:

- control theory,
- numerical linear algebra,
- geometric optimization,
- and data-driven system analysis.

# Selected Subproblem

One of the central subproblems addressed in the article is the following:

> How can we compute the minimum-energy control input for a linear system using only experimental data, without explicitly identifying the system matrices A and B?

This problem represents the core idea of the paper and directly connects:

- controllability,
- experimental data,
- pseudoinverse theory,
- and geometric properties of linear systems.

In the classical model-based formulation, the minimum-energy control input is computed using the controllability matrix:

$$
C_T=
\begin{bmatrix}
B & AB & A^2B & \cdots & A^{T-1}B
\end{bmatrix},
$$

leading to the optimal solution:

$$
u^*=C_T^\dagger x_f.
$$

However, this formulation requires explicit knowledge of the matrices:

$$
A
\quad \text{and} \quad
B.
$$

The article proposes a data-driven alternative based solely on experimental trajectories.

The key idea is to perform multiple control experiments, generating:

- experimental input sequences;
- and their corresponding final states.

These experiments are organized into the matrices:

$$
U=
\begin{bmatrix}
u_1 & \cdots & u_N
\end{bmatrix},
$$

and:

$$
X=
\begin{bmatrix}
x_1 & \cdots & x_N
\end{bmatrix}.
$$

Using the relationship:

$$
X=C_TU,
$$

the article shows that the minimum-energy control can be approximated directly from experimental data through:

$$
\hat u=UX^\dagger x_f.
$$

Geometrically, this formulation reconstructs the optimal control as a linear combination of previously observed experimental inputs.

The quality of this reconstruction depends on how well the experimental inputs span the control signal space:

$$
\mathbb{R}^{mT}.
$$

The following implementation reproduces this data-driven minimum-energy control formulation and compares it against the classical model-based solution.

# Numerical Example

In this example, we consider a discrete-time linear system:

$$
x(t+1)=Ax(t)+Bu(t),
$$

with:

$$
A=
\begin{bmatrix}
0.9 & 0.2 \\
-0.1 & 0.8
\end{bmatrix},
$$

and:

$$
B=
\begin{bmatrix}
1 \\
0.5
\end{bmatrix}.
$$

The objective is to drive the system from the origin:

$$
x_0=
\begin{bmatrix}
0 \\
0
\end{bmatrix},
$$

to the target state:

$$
x_f=
\begin{bmatrix}
1 \\
1
\end{bmatrix},
$$

using a control horizon:

$$
T=4.
$$

We then compare:

1. the classical model-based minimum-energy control;
2. the data-driven control computed only from experimental data.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
A = np.array([
    [0.9,  0.2],
    [-0.1, 0.8]
])

B = np.array([
    [1.0],
    [0.5]
])

n = A.shape[0]
m = B.shape[1]
T = 4

x0 = np.zeros((n,1))
xf = np.array([
    [1.0],
    [1.0]
])

CT_blocks = []
for k in range(T):

    CT_blocks.append(
        np.linalg.matrix_power(A, T-1-k) @ B
    )

CT = np.hstack(CT_blocks)

print("Controllability Matrix:\n")
print(CT)

print("\nRank of CT:")
print(np.linalg.matrix_rank(CT))

u_model = np.linalg.pinv(CT) @ xf

print("\nModel-Based Minimum-Energy Control:\n")
print(u_model)

N = m * T
U = np.random.randn(m*T, N)
X = CT @ U

u_data = U @ np.linalg.pinv(X) @ xf
print("\nData-Driven Control:\n")
print(u_data)

xf_model = CT @ u_model
xf_data = CT @ u_data

error_model = np.linalg.norm(xf - xf_model)
error_data = np.linalg.norm(xf - xf_data)

print("\nModel-Based Final State Error:")
print(error_model)

print("\nData-Driven Final State Error:")
print(error_data)

energy_model = np.linalg.norm(u_model)**2
energy_data = np.linalg.norm(u_data)**2

print("\nModel-Based Control Energy:")
print(energy_model)

print("\nData-Driven Control Energy:")
print(energy_data)
```

```python
N_values = np.arange(1, 15)

errors = []
energies = []

u_model = np.linalg.pinv(CT) @ xf
energy_model = np.linalg.norm(u_model)**2

for N in N_values:
    U = np.random.randn(m*T, N)
    X = CT @ U

    u_data = U @ np.linalg.pinv(X) @ xf  # Data-Driven Solution
    xf_data = CT @ u_data # Final State Reconstruction

    error = np.linalg.norm(xf - xf_data)
    energy = np.linalg.norm(u_data)**2
    
    errors.append(error)
    energies.append(energy)
```

```python
plt.figure(figsize=(10,5))
plt.plot(
    N_values,
    energies,
    marker='o',
    linewidth=2,
    label='Data-Driven'
)
plt.axhline(
    energy_model,
    linestyle='--',
    linewidth=2,
    label='Model-Based Optimal Energy'
)

plt.xlabel("Number of Experiments (N)")
plt.ylabel("Control Energy")
plt.title("Control Energy vs Number of Experiments")

plt.grid(True)
plt.legend()
plt.show()
```

```python
def simulate_system(A, B, u, x0):
    x = x0.copy()
    trajectory = [x.flatten()]

    for k in range(T):

        uk = np.array([[u[k,0]]])
        x = A @ x + B @ uk
        trajectory.append(x.flatten())

    return np.array(trajectory)

traj_model = simulate_system(A, B, u_model, x0) # Model-Based Trajectory
N_list = [1, n, m*T] # Different Experimental Sizes
trajectories = {}

for N in N_list:
    U = np.random.randn(m*T, N)
    X = CT @ U

    u_data = U @ np.linalg.pinv(X) @ xf
    traj_data = simulate_system(A, B, u_data, x0)
    trajectories[N] = traj_data
```

```python
plt.figure(figsize=(10,8))
plt.plot(
    traj_model[:,0],
    traj_model[:,1],
    marker='o',
    linewidth=3,
    label='Model-Based Optimal'
) # Model-Based

markers = ['x', 's', '^']
for i, N in enumerate(N_list):

    traj = trajectories[N]
    plt.plot(
        traj[:,0],
        traj[:,1],
        marker=markers[i],
        linewidth=2,
        label=f'Data-Driven (N={N})'
    ) # Data-Driven

plt.scatter(
    x0[0],
    x0[1],
    s=120,
    label='Initial State'
)

plt.scatter(
    xf[0],
    xf[1],
    s=120,
    label='Target State'
)

plt.xlabel("$x_1$")
plt.ylabel("$x_2$")
plt.title("State-Space Trajectories")

plt.grid(True)
plt.legend()
plt.show()
```

# Numerical Interpretation

The numerical experiments confirm that both the classical model-based formulation and the proposed data-driven methodology are capable of driving the system toward the desired target state.

The controllability matrix obtained for the system is:

$$
C_T=
\begin{bmatrix}
0.892 & 0.96 & 1.00 & 1.00 \\
0.016 & 0.14 & 0.30 & 0.50
\end{bmatrix},
$$

with:

$$
\mathrm{rank}(C_T)=2.
$$

Since the rank of the controllability matrix is equal to the dimension of the state space:

$$
\mathrm{rank}(C_T)=n=2,
$$

the system is controllable over the selected control horizon. Geometrically, this means that the propagated control directions span the entire state space:

$$
\mathbb{R}^2.
$$

Consequently, any target state inside the state space can theoretically be reached through a suitable control sequence.

The classical model-based minimum-energy solution produced the control input:

$$
u^*=
\begin{bmatrix}
-1.0988 \\
-0.4059 \\
0.5523 \\
1.8174
\end{bmatrix},
$$

while the data-driven formulation generated:

$$
\hat u=
\begin{bmatrix}
-2.0461 \\
0.5863 \\
0.9025 \\
1.3598
\end{bmatrix}.
$$

Although the two control vectors are numerically different, both solutions successfully steer the system toward the desired final state.

The model-based reconstruction error was approximately:

$$
2.22\times10^{-16},
$$

which corresponds essentially to machine precision and validates the correctness of the classical optimal solution.

The data-driven formulation also achieved a very small reconstruction error, demonstrating that the experimental dataset contained sufficient geometric information to reconstruct a feasible control action directly from data.

However, an important difference emerges in terms of control energy.

The computed control energies were:

- Model-based energy:

$$
||u^*||_2^2
\approx
4.98;
$$

- Data-driven energy:

$$
||\hat u||_2^2
\approx
7.19.
$$

This discrepancy reveals one of the central theoretical insights of the article.

Although the experimental dataset was sufficient to reach the target state, it was not yet rich enough to perfectly reconstruct the exact orthogonal minimum-energy solution inside the full control signal space:

$$
\mathbb{R}^{mT}.
$$

The additional experiments introduced later provide a deeper understanding of this phenomenon.

The analysis of the reconstruction error and control energy as functions of the number of experiments \(N\) demonstrated that the quality of the data-driven solution improves progressively as the experimental subspace becomes richer.

For small values of:

$$
N,
$$

the experimental inputs do not sufficiently span the control signal space, leading to:

- larger reconstruction errors;
- higher control energy;
- and poorer approximation of the optimal trajectory.

As \(N\) increases, the experimental dataset gradually captures a larger portion of the control space geometry, improving both:

- state reconstruction accuracy;
- and approximation of the minimum-energy solution.

In particular, the experiments numerically validated the theoretical distinction between:

$$
N \ge n
$$

and:

$$
N = mT.
$$

The first condition is associated with state reachability, meaning that the experimental data become sufficient to approximately reach arbitrary states in:

$$
\mathbb{R}^n.
$$

The second condition is associated with exact reconstruction of the optimal minimum-energy control law, since it guarantees sufficient coverage of the entire control signal space:

$$
\mathbb{R}^{mT}.
$$

The state-space trajectory experiments further reinforced this interpretation geometrically.

For small numbers of experiments, the data-driven trajectories deviated significantly from the optimal model-based trajectory, indicating insufficient geometric coverage of the control space.

As the number of experiments increased, the reconstructed trajectories progressively converged toward the optimal trajectory generated by the model-based formulation.

This behavior illustrates one of the deepest geometric insights of the paper:

reaching a desired final state is not equivalent to reconstructing the exact minimum-energy control trajectory associated with that state.

The experiments therefore demonstrate that the performance of the data-driven methodology depends fundamentally on:

- the rank properties of the experimental matrices;
- the richness of the experimental dataset;
- the geometric coverage of the control signal space;
- and the numerical conditioning of the data-driven reconstruction problem.

Overall, the numerical results validate the central idea of the article: minimum-energy control can be reconstructed directly from experimental data without explicit identification of the system matrices, provided that the experimental trajectories contain sufficiently rich geometric information about the underlying controllable subspaces.
